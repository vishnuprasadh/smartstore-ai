```python
from pathlib import Path
import os

PROJECT_ROOT = Path.home() / "documents/personal/AwsLearning/smartstore-ai"
os.chdir(PROJECT_ROOT)

Path("data/raw/openfoodfacts").mkdir(parents=True, exist_ok=True)
Path("data/processed").mkdir(parents=True, exist_ok=True)

print("Working directory:", Path.cwd())
```

```python
import json
import time
import urllib.parse
import urllib.request
from pathlib import Path

BASE = "https://world.openfoodfacts.org/api/v2/search"

FIELDS = [
    "code", "product_name", "generic_name",
    "brands", "categories", "quantity",
    "product_quantity", "product_quantity_unit",
    "packaging", "packaging_text",
    "ingredients_text", "allergens",
    "nutriments", "nutrition_data_per",
    "countries", "image_url", "image_front_url",
    "image_ingredients_url", "image_nutrition_url",
    "stores", "manufacturing_places",
    "labels", "languages_tags",
    "last_modified_t", "data_sources"
]

products = []
seen = set()

TARGET = 1000
PAGE_SIZE = 100
page = 1

while len(products) < TARGET:

    params = {
        "countries_tags_en": "india",
        "page": page,
        "page_size": PAGE_SIZE,
        "fields": ",".join(FIELDS),
        "sort_by": "popularity_key"
    }

    url = BASE + "?" + urllib.parse.urlencode(params)

    request = urllib.request.Request(
        url,
        headers={"User-Agent": "SmartStoreLearning/1.0"}
    )

    try:
        with urllib.request.urlopen(request, timeout=60) as response:
            data = json.load(response)
    except Exception as e:
        print("Download error:", e)
        break

    batch = data.get("products", [])

    if not batch:
        print("No more products available.")
        break

    for product in batch:
        code = product.get("code")

        if code and code not in seen:
            seen.add(code)
            products.append(product)

    print(f"Page {page}: {len(products)} products collected")

    page += 1
    time.sleep(1)

    if page > 30:
        break

output = Path("data/raw/openfoodfacts/products.json")

with output.open("w", encoding="utf-8") as f:
    json.dump(products, f, ensure_ascii=False, indent=2)

print("\nDownload completed")
print("Products:", len(products))
print("File size:", round(output.stat().st_size / (1024 * 1024), 2), "MB")
```

```python
import pandas as pd

df = pd.json_normalize(products)

print("Number of products:", len(df))
print("Number of attributes:", len(df.columns))

display(df.head(5))
```

```python
quality = pd.DataFrame({
    "field": df.columns,
    "populated_records": df.notna().sum().values,
    "completeness_pct": (
        df.notna().mean().values * 100
    ).round(1)
})

display(
    quality.sort_values(
        "completeness_pct",
        ascending=False
    )
)
```
